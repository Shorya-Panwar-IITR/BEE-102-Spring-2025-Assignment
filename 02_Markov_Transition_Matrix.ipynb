```python
import pandas as pd
from collections import defaultdict
```

```python
seq = "CTTCATGTGAAAGCAGACGTAAGTCA"

trans = defaultdict(lambda: defaultdict(int))

for i in range(len(seq) - 1):
    curr = seq[i]
    nxt = seq[i + 1]
    trans[curr][nxt] += 1

nucleotides = ['A', 'C', 'G', 'T']
transitionMatrix = pd.DataFrame(index=nucleotides, columns=nucleotides)

for curr in nucleotides:
    currSum = sum(trans[curr].values())
    for nxt in nucleotides:
        if currSum > 0:
            transitionMatrix.loc[curr, nxt] = trans[curr][nxt] / currSum
        else:
            transitionMatrix.loc[curr, nxt] = 0.0

print("Markov Transition Matrix:")
display(transitionMatrix)
```
