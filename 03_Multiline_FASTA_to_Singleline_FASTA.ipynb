```python
inputFile = open('./multiline_input.fasta', 'r')
outputFile = open('./singleline_output.fasta', 'w')

firstSeq = True

for line in inputFile:
    if line.startswith(">"):
        if firstSeq == False:
            outputFile.write("\n")
        else:
            firstSeq = False
        outputFile.write(line.strip() + "\n")
    else:
        outputFile.write(line.strip())

outputFile.write("\n")

inputFile.close()
outputFile.close()

print('Output file created: singleline_output.fasta')
```
